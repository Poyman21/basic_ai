## 2 · list, dict และ tuple (15 นาที)

| ชนิด | หน้าตา | ใช้เมื่อ | ตัวอย่างใน introduction.ipynb |
|---|---|---|---|
| list | `[a, b, c]` | เก็บหลายค่าเรียงตามลำดับ เพิ่มหรือลดได้ | `TOKEN_FIELDS`, `model_rows` |

````python
TOKEN_FIELDS = ["input", "output", "reasoning", "total"]

print(TOKEN_FIELDS[0])     # ตำแหน่งเริ่มนับที่ 0 ได้ input
print(TOKEN_FIELDS[-1])    # -1 คือตัวสุดท้าย ได้ total
print(len(TOKEN_FIELDS))   # มีกี่ตัว ได้ 4

model_rows = []                       # list ว่าง
model_rows.append("แถวของ Sonnet")    # .append() เพิ่มต่อท้าย
model_rows.append("แถวของ Haiku")
print(model_rows)
````

````python
usage = {
    "prompt_tokens": 180,
    "completion_tokens": 3,
    "total_tokens": 183,
}

print(usage["prompt_tokens"])            # ดึงค่าด้วยกุญแจ ได้ 180
print(usage.get("reasoning_tokens"))     # .get() ถ้าไม่มีกุญแจนี้จะได้ None โดยไม่ error
print(usage.get("reasoning_tokens", 0))  # ใส่ค่าสำรองเองได้ ได้ 0

usage["finish"] = "stop"                 # เพิ่มหรือแก้ค่า
print(usage)
print(list(usage.keys()))                # ดูว่ามีกุญแจอะไรบ้าง
````

## 3 · for และ if (10 นาที)

**ใน Python การย่อหน้าเป็นส่วนหนึ่งของไวยากรณ์** บรรทัดที่ลงท้ายด้วย `:` ต้องตามด้วยบรรทัดที่ย่อหน้าเข้าไป 4 ช่อง และบรรทัดที่ย่อหน้าเท่ากันถือว่าอยู่กลุ่มเดียวกัน JupyterLab จะย่อหน้าให้เองเมื่อกด Enter

````python
models = [
    ("Claude Sonnet 4.6", "SONNET-ID"),
    ("Claude Haiku 4.5", "HAIKU-ID"),
]

for name, model_id in models:        # วนทีละคู่ แล้วแตก tuple ใส่ name กับ model_id
    print("กำลังทดสอบ", name)          # ย่อหน้าแปลว่าอยู่ใน loop ทำซ้ำทุกรอบ
    print("    model id:", model_id)
print("จบ loop")                       # ไม่ย่อหน้าแปลว่าอยู่นอก loop ทำครั้งเดียว
````

## 4 · function (20 นาที)

function คือ "สูตร" ที่ตั้งชื่อไว้ รับค่าเข้า (parameter) ทำงาน แล้วส่งผลกลับด้วย `return`
- `def ชื่อ(...):` คือการ **นิยาม** สูตร ตอนนี้ยังไม่ทำงาน
- `ชื่อ(...)` คือการ **เรียกใช้** สูตร ตอนนี้ทำงานจริง

````python
def add_tokens(input_tokens, output_tokens):
    """รวม token ขาเข้าและขาออก"""           # docstring คำอธิบาย function
    total = input_tokens + output_tokens
    return total                              # ส่งค่ากลับให้ผู้เรียก

result = add_tokens(180, 3)                           # ส่งค่าเรียงตามลำดับ
print(result)
print(add_tokens(output_tokens=3, input_tokens=180))  # keyword argument ระบุชื่อ ลำดับจึงไม่สำคัญ
````

**ลองเอง:** เพิ่ม cell ใหม่ พิมพ์ `add_tokens(` แล้วกด `Shift + Tab` หรือรัน `help(add_tokens)` จะเห็นว่า function รับค่าอะไรและมี docstring ว่าอะไร วิธีนี้ใช้ได้กับทุก function ใน introduction.ipynb เช่น `help(chat)`

````python
def describe(model, budget=1024):           # budget มีค่าเริ่มต้น จึงไม่ส่งมาก็ได้
    return f"{model} คิดได้ไม่เกิน {budget} tokens"

print(describe("Haiku"))
print(describe("Haiku", budget=2048))
````

รูปแบบเดียวกับ `os.getenv("AI_ROUTER_URL", "http://localhost:8080/v1")` ค่าที่สองคือค่าสำรองเมื่อหาค่าที่ตั้งไว้ไม่เจอ

````python
def measure_demo(label):
    row = {"การตั้งค่า": label, "total": 183}
    reply_text = "64"
    return row, reply_text        # ส่งกลับ 2 ค่าพร้อมกัน

row, reply_text = measure_demo("Claude Haiku 4.5")   # รับ 2 ค่าด้วย 2 ตัวแปร
print(row)
print(reply_text)
````

### `**settings` ใน `chat()`

ใน introduction.ipynb มี `def chat(model_id, **settings):` ซึ่ง `**settings` แปลว่า "รับ keyword argument เพิ่มกี่ตัวก็ได้ แล้วรวมเก็บเป็น dict ชื่อ `settings`" ลองดู function จำลองด้านล่าง

````python
def fake_chat(model_id, **settings):
    print("model_id =", model_id)
    print("settings =", settings)
    return {"model": model_id, "max_retries": 0, "timeout": 90, **settings}

fake_chat("HAIKU-ID", max_completion_tokens=2048, extra_body={"thinking": {"type": "disabled"}})
````

ดังนั้น `chat(HAIKU, max_completion_tokens=2048, extra_body={"thinking": thinking})` ในหัวข้อ thinking คือการส่งค่าตั้งทั้งสองตัวต่อไปให้ `ChatOpenAI(...)`

````python
print(add_tokens)          # ไม่มี () แค่อ้างถึง function ยังไม่ได้เรียกใช้ จะเห็น <function ...>
print(add_tokens(1, 2))    # มี () คือเรียกใช้จริง ได้ 3
````

ถ้าเห็น `<function ... at 0x...>` หรือ `<bound method ...>` แทนผลลัพธ์ แปลว่าลืมใส่ `()`

````python
def cycle_minutes(machine: str) -> int:
    """Look up the duration in minutes of one complete production cycle for machine A or B."""
    return {"A": 8, "B": 12}[machine]

print(cycle_minutes("A"))
print(cycle_minutes("B"))
````

- `machine: str` และ `-> int` เรียกว่า **type hint** เป็นป้ายบอกว่าควรส่งข้อความเข้าและจะได้จำนวนเต็มออก Python ไม่บังคับ แต่ช่วยให้คนอ่านและ AI เข้าใจ
- ใน introduction.ipynb มี `@tool` อยู่บรรทัดบน `def` เรียกว่า **decorator** มันห่อ function ธรรมดาให้เป็นเครื่องมือที่โมเดล AI ขอเรียกใช้ได้ และส่ง **ชื่อ function, type hint และ docstring** ไปให้โมเดลอ่าน docstring จึงสำคัญมาก
- function ที่ห่อด้วย `@tool` แล้วจะเรียกผ่าน `.invoke(...)` เช่น `first_finish_minute.invoke({"target_units": 460})`

````python
from types import SimpleNamespace

# จำลองหน้าตาของ reply = model.invoke(PROMPT) ใน introduction.ipynb (ตัวเลขสมมติ ของจริงต้องต่อ AI Router)
reply = SimpleNamespace(
    content="64",
    tool_calls=[],
    response_metadata={
        "finish_reason": "stop",
        "token_usage": {"prompt_tokens": 180, "completion_tokens": 3, "total_tokens": 183},
    },
)

print(reply.content)                                  # attribute
print(reply.response_metadata["finish_reason"])       # attribute ที่เป็น dict แล้วดึงค่าด้วยกุญแจ
print(reply.response_metadata.get("token_usage"))     # attribute แล้วเรียก method .get()
````

### 6.4 error ที่ไม่ขึ้นสีแดง: `raise` และ `try` / `except`

- `raise` คือผู้เขียน function **ตั้งใจโยน error** พร้อมข้อความบอกวิธีแก้
- `try` / `except` คือ **ดักจับ error** แล้วทำอย่างอื่นแทน cell จึงไม่หยุดกลางทาง

function ด้านล่างคัดลอกมาจาก introduction.ipynb (ตัด `@tool` ออก)

````python
def first_finish_minute(target_units: int) -> int:
    """Return the earliest minute the two workshop machines finish the target units."""
    if not 1 <= target_units <= 100_000:
        raise ValueError("target_units must be between 1 and 100000")
    # เครื่อง A ทำงานตลอด จึงใช้เวลาของ A เพียงเครื่องเดียวเป็นขอบบนได้
    last_minute = 8 * ((target_units + 35) // 36)
    for minute in range(last_minute + 1):
        b_minutes = min(minute, 24) + max(0, minute - 36)
        if 36 * (minute // 8) + 50 * (b_minutes // 12) >= target_units:
            return minute
    raise RuntimeError("Production target was not reached")

print(first_finish_minute(460))   # ได้ 64 ตรงกับเฉลยใน introduction.ipynb
````

````python
first_finish_minute(0)   # อ่านบรรทัดสุดท้าย ผู้เขียนบอกไว้แล้วว่าต้องส่งค่าเท่าไร
````

````python
for target in [460, 0, 1000]:
    try:                                          # ลองทำ
        minute = first_finish_minute(target)
        print(f"{target} ชิ้น → นาทีที่ {minute}")
    except ValueError as exc:                     # ถ้าเกิด ValueError ให้ทำส่วนนี้แทน และวนรอบถัดไปต่อ
        print(f"{target} ชิ้น → ข้าม: {type(exc).__name__}: {exc}")
````

cell ด้านล่างคัดลอกส่วนแปลงผลลัพธ์และ `show_comparison()` มาจาก cell 6 **โดยตัดส่วนที่ต้องต่อ AI Router ออก** แล้วป้อนคำตอบสมมติแทน ลองอ่านทีละบรรทัดว่าใช้เรื่องอะไรที่เรียนมาบ้าง

````python
import pandas as pd
from IPython.display import display

TOKEN_FIELDS = ["input", "output", "reasoning", "total"]


def parse_reply(label, reply):
    row = {"การตั้งค่า": label, **dict.fromkeys(TOKEN_FIELDS)}
    usage = reply.response_metadata.get("token_usage") or {}
    details = usage.get("completion_tokens_details") or {}
    row.update(
        input=usage.get("prompt_tokens"),
        output=usage.get("completion_tokens"),
        reasoning=details.get("reasoning_tokens"),
        total=usage.get("total_tokens"),
        finish=reply.response_metadata.get("finish_reason"),
    )
    calls = [call["name"] for call in reply.tool_calls]
    row["คำตอบ / สถานะ"] = "ขอใช้ " + ", ".join(calls) if calls else reply.content
    return row


def show_comparison(rows):
    table = pd.DataFrame(rows).set_index("การตั้งค่า")
    table[TOKEN_FIELDS] = table[TOKEN_FIELDS].astype("Int64")
    for field in ("input", "total"):
        table[f"Δ {field}"] = table[field] - table[field].iloc[0]
    columns = [*TOKEN_FIELDS, "Δ input", "Δ total", "finish", "คำตอบ / สถานะ"]
    table = table.reindex(columns=columns)
    display(table.astype(object).where(table.notna(), "—"))
````

````python
from types import SimpleNamespace


# ตัวเลขสมมติ ใช้ดูหน้าตาตารางโดยไม่ต้องต่อ AI Router
def fake_reply(content, prompt_tokens, completion_tokens, reasoning_tokens=None, tool_calls=()):
    usage = {
        "prompt_tokens": prompt_tokens,
        "completion_tokens": completion_tokens,
        "total_tokens": prompt_tokens + completion_tokens,
    }
    if reasoning_tokens is not None:
        usage["completion_tokens_details"] = {"reasoning_tokens": reasoning_tokens}
    return SimpleNamespace(
        content=content,
        tool_calls=list(tool_calls),
        response_metadata={
            "finish_reason": "tool_calls" if tool_calls else "stop",
            "token_usage": usage,
        },
    )


tool_call = {"name": "first_finish_minute", "args": {"target_units": 460}, "id": "call_1"}

rows = [
    parse_reply("Haiku ปิด thinking", fake_reply("64", 180, 4)),
    parse_reply("Haiku เปิด thinking", fake_reply("64", 180, 640, reasoning_tokens=600)),
    parse_reply("Haiku + 1 tool", fake_reply("", 430, 70, tool_calls=[tool_call])),
]
show_comparison(rows)
````
